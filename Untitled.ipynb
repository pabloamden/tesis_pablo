```python
from pathlib import Path
import pandas as pd
import re

# Cambiá esta ruta por la tuya
root_dir = Path("Corpus")

def detectar_corpus(path: Path):
    path_str = str(path).lower().replace("\\", "/")

    if "transcripciones-segunda-vuelta/highvol" in path_str:
        return "segunda-vuelta/highVol"
    elif "transcripciones-segunda-vuelta/random" in path_str:
        return "segunda-vuelta/random"
    elif "raw_ros_hv" in path_str:
        return "raw_ROS_HV"
    elif "raw_ros" in path_str:
        return "raw_ROS"
    else:
        return "desconocido"

def extraer_chico(nombre_archivo: str):
    """
    Ejemplos:
    jeremiasc-a1-nsm.csv  -> jeremiasc
    jeremiasc-a2-nsb.rttm -> jeremiasc
    """
    nombre_base = Path(nombre_archivo).stem.lower()

    m = re.match(r"^(.*?)-a\d+", nombre_base)
    if m:
        return m.group(1)

    # fallback por si no matchea el patrón
    return nombre_base

rows = []

for file in root_dir.rglob("*"):
    if file.is_file():
        chico = extraer_chico(file.name)
        corpus = detectar_corpus(file)

        rows.append({
            "chico": chico,
            "archivo": file.name,
            "corpus": corpus,
            "path": str(file)
        })

df = pd.DataFrame(rows)

# Ver todas las asociaciones archivo -> chico -> corpus
print(df.head())

# Quedarse con chico -> corpus
df_chicos = (
    df[["chico", "corpus"]]
    .drop_duplicates()
    .sort_values(["chico", "corpus"])
    .reset_index(drop=True)
)

print("\nTabla chico -> corpus:")
print(df_chicos)

# Detectar chicos que aparecen en más de un corpus
df_conflictos = (
    df_chicos.groupby("chico", as_index=False)
    .agg(n_corpus=("corpus", "nunique"),
         corpus_list=("corpus", lambda x: ", ".join(sorted(x.unique()))))
)

df_conflictos = df_conflictos[df_conflictos["n_corpus"] > 1]

print("\nChicos con más de un corpus:")
print(df_conflictos)

# Guardar resultados

df_chicos.to_csv("chico_corpus.csv", index=False)
```
